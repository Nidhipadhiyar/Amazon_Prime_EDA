## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
titles_clean = titles.copy()
credits_clean = credits.copy()
````

````python
titles_clean = titles_clean.drop_duplicates()
credits_clean = credits_clean.drop_duplicates()
````

````python
print("Duplicates in titles:", titles_clean.duplicated().sum())
print("Duplicates in credits:", credits_clean.duplicated().sum())
````

````python
titles_clean['type'].value_counts()
````

````python
shows = titles_clean[titles_clean['type'] == 'SHOW'].copy()
````

````python
import ast

titles_clean['genres'] = titles_clean['genres'].apply(ast.literal_eval)
titles_clean['production_countries'] = titles_clean['production_countries'].apply(ast.literal_eval)
````

````python
merged = titles_clean.merge(credits_clean,on='id',how='left')
````

#### Chart - 14 - Correlation Heatmap

Most other variables have correlations close to 0, showing little or no strong linear relationship. Correlation values closer to ±1 indicate stronger relationships, while values near 0 indicate weak relationships
